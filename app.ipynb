```python
import sys, pathlib
# The model was exported on Python 3.13. On older Pythons, map the 3.13-only module name.
if not hasattr(pathlib, '_local'):
    sys.modules['pathlib._local'] = pathlib

from fastai.vision.all import *
import ipywidgets as widgets

learn_inf = load_learner('export.pkl')

btn_upload = widgets.FileUpload()
btn_run = widgets.Button(description='Classify')
out_pl = widgets.Output()
lbl_pred = widgets.Label()

def on_click_classify(change):
    img = PILImage.create(bytes(btn_upload.value[-1]['content']))
    out_pl.clear_output()
    with out_pl:
        display(img.to_thumb(128, 128))
    pred, pred_idx, probs = learn_inf.predict(img)
    lbl_pred.value = f'Prediction: {pred}; Probability: {probs[pred_idx]:.4f}'

btn_run.on_click(on_click_classify)
widgets.VBox([widgets.Label('Upload a bear photo'), btn_upload, btn_run, out_pl, lbl_pred])
```
